Goal: find these values to maximize profit
- x1 = optimal number of wheat hectares (answer: 6)
- x2 = optimal number of barley hectares (answer: 4)

Given these parameters:
- S1 = price of wheat = 4
- S2 = price of barley = 3
- L = total farm land = 10
- F = max kg fertilizer available
- F1 = how much kg of fertilizer a hectare of wheat requires = 2
- F2 = how much kg of fertilizer a hectare of barley requires = 4
- P = max kg pesticide available
- P1 = how much kg of pesticide a hectare of wheat requires = 6
- P2 = how much kg of pesticide a hectare of barley requires = 3

Mathematically:
- maximize [S1, S2] * [[x1], [x2]]
- subject to [[1, 1], [F1, F2], [P1, P2]] * [[x1], [x2]] <= [[L], [F], [P]]
- and [[x1], [x2]] >= [[0], [0]]

Below, allow:
- c = [[x1], [x2]] = coefficients to minimize
- A_ub = [[1, 1], [F1, F2], [P1, P2]] = constraint matrix
- b_ub = [[10], [32], [48]] = constraint vector (upper bounds)

https://en.wikipedia.org/wiki/Linear_programming

```python
import numpy as np
from scipy.optimize import linprog
```

```python
c = [-4, -3] # negated to maximize
L = 10

A_ub = np.array([[1,1], [2,4], [6,3]])
b_ub = np.array([10, 32, 48])
bounds = [(0, None), (0, None)]

results = linprog(c=c, A_ub=A_ub, b_ub=b_ub)
print(results.success, results.message)
```

```python
# solution
print(results.x)
```

```python
# the resulting profit
print(-results.fun)  # negated to maximize
```
